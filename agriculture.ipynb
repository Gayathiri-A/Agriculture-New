```python
import pandas as pd
```

```python
df= pd.read_csv("C:\\Users\\Gayathiri A\\Downloads\\ICRISAT-District Level Data - ICRISAT-District Level Data.csv")
```

```python
df.dropna(axis=1, how='all',inplace = True)
```

```python
df.fillna("unknown",inplace = True)
```

```python
import matplotlib.pyplot as plt
```

```python
plt.figure(figsize=(10,6))
plt.plot(df['Year'], df['RICE PRODUCTION (1000 tons)'], marker='o', label='Rice')
plt.title('Rice Production in Durg (1966–1974)')
plt.xlabel('Year')
plt.ylabel('Production (1000 tons)')
plt.grid(True)
plt.legend()
plt.show()

```

```python
plt.figure(figsize=(10,6))
plt.plot(df['Year'], df['WHEAT PRODUCTION (1000 tons)'], marker='s', color='orange', label='Wheat')
plt.title('Wheat Production in Durg (1966–1974)')
plt.xlabel('Year')
plt.ylabel('Production (1000 tons)')
plt.grid(True)
plt.legend()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.plot(df['Year'], df['RICE PRODUCTION (1000 tons)'], label='Rice')
plt.plot(df['Year'], df['WHEAT PRODUCTION (1000 tons)'], label='Wheat')
plt.title('Rice vs Wheat Production (1966–1974)')
plt.xlabel('Year')
plt.ylabel('Production (1000 tons)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.plot(df['Year'], df['SUGARCANE PRODUCTION (1000 tons)'], marker='^', color='green')
plt.title('Sugarcane Production in Durg (1966–1974)')
plt.xlabel('Year')
plt.ylabel('Production (1000 tons)')
plt.grid(True)
plt.show()
```

```python
df['Total Millet'] = df['PEARL MILLET PRODUCTION (1000 tons)'] + df['FINGER MILLET PRODUCTION (1000 tons)']
plt.figure(figsize=(10,6))
plt.plot(df['Year'], df['Total Millet'], marker='d', color='purple')
plt.title('Millet Production in Durg (1966–1974)')
plt.xlabel('Year')
plt.ylabel('Production (1000 tons)')
plt.grid(True)
plt.show()
```

```python
import numpy as np

bar_width = 0.35
years = df['Year']
index = np.arange(len(years))

plt.figure(figsize=(12,6))
plt.bar(index, df['KHARIF SORGHUM PRODUCTION (1000 tons)'], bar_width, label='Kharif')
plt.bar(index + bar_width, df['RABI SORGHUM PRODUCTION (1000 tons)'], bar_width, label='Rabi')
plt.xlabel('Year')
plt.ylabel('Production (1000 tons)')
plt.title('Sorghum Production (Kharif vs Rabi)')
plt.xticks(index + bar_width / 2, years)
plt.legend()
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.scatter(df['RICE AREA (1000 ha)'], df['RICE PRODUCTION (1000 tons)'], label='Rice', alpha=0.7)
plt.scatter(df['WHEAT AREA (1000 ha)'], df['WHEAT PRODUCTION (1000 tons)'], label='Wheat', alpha=0.7)
plt.title('Area vs Production (Rice & Wheat)')
plt.xlabel('Area (1000 ha)')
plt.ylabel('Production (1000 tons)')
plt.legend()
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(10,6))
plt.plot(df['Year'], df['RICE YIELD (Kg per ha)'], label='Rice Yield')
plt.plot(df['Year'], df['WHEAT YIELD (Kg per ha)'], label='Wheat Yield')
plt.title('Yield Comparison (Rice vs Wheat)')
plt.xlabel('Year')
plt.ylabel('Yield (Kg per ha)')
plt.legend()
plt.grid(True)
plt.show()
```
